```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    "Branch": ["CSE", "ECE", "Electrical", "Mechanical", "Civil", "Chemical"],
    "Students": [120, 100, 90, 110, 80, 70],
    "Placed": [115, 90, 78, 85, 55, 52],
    "Average_Package": [24, 16, 12, 10, 7, 9],
    "Highest_Package": [80, 45, 35, 28, 18, 22]
}
```

```python
df = pd.DataFrame(data)
```

```python
total_students = np.sum(df["Students"])
print("Total Students:", total_students)
```

```python
total_placed = np.sum(df["Placed"])
print("Total Placed Students:", total_placed)
```

```python
unplaced = total_students - total_placed
print("Unplaced Students:", unplaced)
```

```python
placement_percentage = (total_placed / total_students) * 100

print("Overall Placement:", placement_percentage, "%")
```

```python
df["Placement_Percentage"] = (df["Placed"] / df["Students"]) * 100

print(df)
```

```python
average_salary = np.mean(df["Average_Package"])

print("Average Package:", average_salary, "LPA")
```

```python
highest_salary = np.max(df["Highest_Package"])

print("Highest Package:", highest_salary, "LPA")
```

```python
branch = df.loc[df["Highest_Package"].idxmax(), "Branch"]

print("Top Branch:", branch)
```

```python
print(df[["Branch", "Students", "Placed", "Placement_Percentage"]])
```

```python
plt.bar(df["Branch"], df["Students"], label="Total Students")
plt.bar(df["Branch"], df["Placed"], label="Placed Students")

plt.xlabel("Branch")
plt.ylabel("Students")
plt.title("IIT Branch Placement Report")

plt.legend()
plt.show()
```

```python
plt.bar(df["Branch"], df["Average_Package"])

plt.xlabel("Branch")
plt.ylabel("Average Package (LPA)")
plt.title("Branch Wise Average Package")

plt.show()
```

```python
plt.bar(df["Branch"], df["Highest_Package"])

plt.xlabel("Branch")
plt.ylabel("Highest Package (LPA)")
plt.title("Branch Wise Highest Package")

plt.show()
```

```python
plt.bar(df["Branch"], df["Placement_Percentage"])

plt.xlabel("Branch")
plt.ylabel("Placement Percentage")
plt.title("Branch Wise Placement Percentage")

plt.show()
```

```python
top_branch = df.loc[df["Placement_Percentage"].idxmax(), "Branch"]

print("Best Placement Branch:", top_branch)
```
